 # 2.Missing Value Handling

```python
df = pd.read_csv("dirty_cafe_sales.csv")
df=df.replace(["ERROR","Error"],np.nan)
#check missing values
print("\nNulls before\n",df.isnull().sum())
#for numeric columns
for col in["Quantity","Price Per Unit","Total Spent"]:
    df[col]=pd.to_numeric(df[col],errors="coerce")
    df[col]=df[col].fillna(df[col].median())
#for categorial columns
for col in["Item","Location","Payment Method"]:
    df[col]=df[col].fillna(df[col].mode()[0])
df["Transaction Date"]=pd.to_datetime(df["Transaction Date"],errors="coerce")
median_date=df["Transaction Date"].dropna().median()
df["Transaction Date"]=df["Transaction Date"].fillna(median_date)

#check after handling
print("\nNull after\n",df.isnull().sum())

```

Numeric missing values were filled using median imputation.
Categorical missing values were replaced with "Unknown".
This helps prepare the dataset for further analysis.

# 3.Duplicate Removal

```python
duplicate_count=df.duplicated().sum()
print("Duplicate_Rows:",duplicate_count)
print(df[df.duplicated(keep=False)])
df = df.drop_duplicates().reset_index(drop=True)
print("After Duplicates Removal:",df.duplicated().sum())
```

Duplicate rows were identified using duplicated().
The dataset contained 0 duplicate rows.
No duplicate rows were removed.

# 4.Standardization

```python
df[["Item","Location","Payment Method"]].head()
for col in ["Item","Location","Payment Method"]:
    df[col]=df[col].astype("str").str.title().str.strip()
print("Unique Item After Standardization",df["Item"].unique())
print("Unique Payment Methods",df["Payment Method"].unique())
print("Unique Loaction:",df["Location"].unique()[:5])
```

```python
# Fix UNKNOWN values - replace with Most Frequent (Mode)
df['Item'] = df['Item'].replace(['UNKNOWN', 'Unknown', 'unknown'], pd.NA)
df['Payment Method'] = df['Payment Method'].replace(['UNKNOWN', 'Unknown', 'unknown'], pd.NA)
df['Location'] = df['Location'].replace(['UNKNOWN', 'Unknown', 'unknown'], pd.NA)

df['Item'] = df['Item'].fillna(df['Item'].mode()[0])
df['Payment Method'] = df['Payment Method'].fillna(df['Payment Method'].mode()[0])
df['Location'] = df['Location'].fillna(df['Location'].mode()[0])

print("After fixing Unknown:")
print(df['Item'].unique())
print(df['Payment Method'].unique())
print(df['Location'].unique())
```

# 5. Outlier Detection

```python
for col in["Quantity","Price Per Unit","Total Spent"]:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3-Q1
    lower=Q1-1.5*IQR
    upper=Q3+1.5*IQR
    outliers=df[(df[col]<lower)|(df[col]>upper)]
    print(col,"Outliers:",len(outliers))
```

Outliers detected using IQR Method:
`Lower = Q1 - 1.5*IQR`, `Upper = Q3 + 1.5*IQR`
 Column : Outlier Count 
 Quantity : 0 
 Price Per Unit : 0 
 Total Spent : 259 
No outliers in Quantity and Price Per Unit. 259 values in Total Spent flagged as outliers but are valid high-value transactions and retained.

# 6.Data Type Correction

```python
print("--- DATA TYPE CORRECTION ---")
df["Transaction ID"] = df["Transaction ID"].astype(str)
df["Transaction Date"] = pd.to_datetime(df["Transaction Date"], errors='coerce')
df["Quantity"] = df["Quantity"].astype(int)
df["Price Per Unit"] = df["Price Per Unit"].astype(float).round(2)
df["Total Spent"] = df["Total Spent"].astype(float).round(2)
df["Item"] = df["Item"].astype(str)
df["Payment Method"] = df["Payment Method"].astype(str)
df["Location"] = df["Location"].astype(str)
print("\nAFTER - Corrected:")
print(df.dtypes)
print(df.head(10))
```
